```python
sudoku_list = [[0, 0, 4, 3, 0, 0, 2, 0, 9],
                [0, 0, 5, 0, 0, 9, 0, 0, 1],
                [0, 7, 0, 0, 6, 0, 0, 4, 3],
                [0, 0, 6, 0, 0, 2, 0, 8, 7],
                [1, 9, 0, 0, 0, 7, 4, 0, 0],
                [0, 5, 0, 0, 8, 3, 0, 0, 0],
                [6, 0, 0, 0, 0, 0, 1, 0, 5],
                [0, 0, 3, 5, 0, 8, 6, 9, 0],
                [0, 4, 2, 9, 1, 0, 3, 0, 0]]
```

```python
sudoku_solution = np.array([[8, 6, 4, 3, 7, 1, 2, 5, 9],
                            [3, 2, 5, 8, 4, 9, 7, 6, 1],
                            [9, 7, 1, 2, 6, 5, 8, 4, 3],
                            [4, 3, 6, 1, 9, 2, 5, 8, 7],
                            [1, 9, 8, 6, 5, 7, 4, 3, 2],
                            [2, 5, 7, 4, 8, 3, 9, 1, 6],
                            [6, 8, 9, 7, 3, 4, 1, 2, 5],
                            [7, 1, 3, 5, 2, 8, 6, 9, 4],
                            [5, 4, 2, 9, 1, 6, 3, 7, 8]])
sudoku_game = np.array(sudoku_list)
print("suduko_game:\n",sudoku_game)
print("suduko_solution:\n",sudoku_solution)
```

```python
plt.scatter(sudoku_game, sudoku_solution)
plt.show()
```

```python
game_and_solution = np.array([sudoku_game, sudoku_solution])

print(game_and_solution)
```
