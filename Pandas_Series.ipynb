# PANDA SERIES

```python
import pandas as pd 
from pandas import Series
```

```python
sales = Series([120,150,180,210])
print(sales.values)
sales.index
```

```python
temp = Series([28,19,3,24],index=['mon','tue','wed','thu'])
temp
```

```python
population = Series([35, 42, 31], index=['Delhi','Mumbai','Jaipur'])
population['Jaipur']
```

```python
score = Series([35, 68, 44, 79, 52])
score[score > 50]
```

```python
subjects = Series([76, 83, 91], index=['Physics','Chemistry','Maths'])
'Chemistry' in subjects.index
```

```python
s = Series([1,2,4,5])
s.to_dict()
```

```python
prices = {'Pen': 10, 'Notebook': 55, 'Bag': 700}
a = Series(prices)
a
```

```python
data = {'A': 10, 'B': 20, 'C': 30}
s = Series(data, index=['A','B','C','D'])
print(pd.isnull(s))
pd.notnull(s)
```

```python
sales.name = 'Monthly Sales'
sales
```

```python
sales.index.name = 'Month'
sales
```

```python
s = Series([18, 45, 27, 61],
           index=['P','Q','R','S'])

s[s > 30]
```

```python
s = Series([500, 600, 700],
           index=['East','West','North'])

print('West' in s)
500 in s
```

```python
city_data = {'Delhi': 80, 'Pune': 65, 'Kochi': 70}

city_series = Series(city_data,
                     index=['Delhi', 'Kochi', 'Pune', 'Chennai'])
city_series
```

```python
s1 = Series([10, 20, 30],
            index=['A','B','C'])

s2 = Series([1, 2, 3],
            index=['C','A','B'])

s1 + s2
```

```python
s1 = Series({'A': 12, 'B': 18})
s2 = Series({'B': 5, 'C': 9})

s1 + s2
```

```python
stock = Series([25, 40, 18],
               index=['Item1','Item2','Item3'])

stock_dict = stock.to_dict()
print(stock_dict)

stock2 = Series(stock_dict)
stock2
```

```python
marks = {'Maths': 88, 'Science': 74,
         'English': 69, 'History': 81}

marks_series = Series(marks,
                      index=['History', 'English', 'Science', 'Maths'])

marks_series[marks_series > 70]
```

```python
north = Series({'Jan': 40, 'Feb': 55, 'Mar': 60})

south = Series({'Feb': 20, 'Mar': 25, 'Apr': 30})

total = north + south
total
```
