#  Conditions — if / elif / else, modulo (%)

```python
for i in range(0, 5): 
    if i % 2 == 0: 
        print(i) 
    else: 
        print(i + 10)
```

```python
for i in range(1,16):
    if i%3 == 0 and i%5 == 0:
        print(i,": fizzbuzz")
    elif i%3 == 0:
        print(i,': fizz')
    elif i%5 == 0:
        print(i,': buzz')
    
```

```python
n = int(input('Enter a number'))
if n==0:
    print('Zero')
elif n>0:
    print('Positive')
elif n<0:
    print('Negative')
```

#  Functions & Passing Arguments

```python
def absolute(val): 
    if val < 0: 
        return 0 - val 
    return val

print(absolute(-8))
```

```python
def print_abs(val):
    if val <0:
         print(0-val)
    else:
         print(val)

print_abs(-2.4)
```

```python
def inc_val(val): 
       val = val + 1 
x = 7 
inc_val(x) 
print(x)
```

```python
def swap(a,b):
    a+=b
    b=a-b
    a=a-b
    print(a,b)
a=1
b=2
swap(1,2)
print(a,b)
```

#  Scope — Local vs Global Variables

```python
def my_abs(val): 
    if val < 0: 
        return 0 - val 
    return val 
print(val)     # NameError: name 'val' is not defined 
# val only exists inside my_abs() 
```

```python
my_val = 0     # global variable 
def show():     # val only exists inside my_abs() 
    print(my_val)   # can READ a global from inside a function
show()
```
